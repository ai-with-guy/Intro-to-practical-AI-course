```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,4))
```

```python
x = np.linspace(0, 5, 11)  
```

```python
axes[0].plot(x, x**2)

axes[0].set_title("X^2")
axes[0].set_xlabel('x')
fig
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,4))
      
axes[0].plot(x, x**2, x, np.exp(x))

axes[0].set_title("Normal scale")
axes[0].set_xlabel('x')

axes[1].plot(x, x**2, x, np.exp(x))
axes[1].set_yscale("log")
axes[1].set_title("Log scale");
axes[1].set_xlabel('x')

axes[0].text(1.5, 60, r"$y=exp(x)$", fontsize=20, color="orange")
axes[1].text(2.5, 1, r"$y=x^2$", fontsize=20, color="blue")
```

## Video 1.3.2

```python
import seaborn as sns
penguins = sns.load_dataset("penguins")

penguins.head()
```

```python
penguins.info()
```

```python
penguins.columns
```

```python
penguins["island"].unique()  # How many unique islands?
```

```python
penguins = penguins.dropna()
penguins.info()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(penguins["body_mass_g"], bins=25, color="steelblue", edgecolor="white")

ax.set_title("Body mass distribution — all penguins")
ax.set_xlabel("Body mass (g)")
ax.set_ylabel("Count")
plt.show()
```

```python
sns.histplot(
    data=penguins,
    x="body_mass_g",
    hue="species",
    stat="count",       # this is actually the default
    kde=True,            # optional: adds a density-shaped curve on top
    palette=('DarkOrange', 'MediumOrchid', 'Teal')
)
```

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g", hue="sex", split=True)
plt.title("Body mass by species and sex")
plt.show()
```

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill dimensions by species")
plt.show()
```

```python
sns.jointplot(data=penguins, x="flipper_length_mm", y="bill_length_mm", hue="species")
```

```python
numeric_cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
corr = penguins[numeric_cols].corr()

plt.figure(figsize=(5, 4))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="vlag", vmin=-1, vmax=1, center=0, square=True)
plt.title("Correlation matrix — penguin measurements")
plt.show()
```

```python
sns.pairplot(penguins, vars=["bill_length_mm", "flipper_length_mm", "body_mass_g"], hue="species")
plt.show()
```

```python
sns.catplot(data=penguins, x="species", hue="sex", col="island", kind="count", palette="Set2")
plt.show()
```
