```python
# create a readable category instead of raw class numbers

df["Pclass_label"] = df["Pclass"].map({1: "First", 2: "Second", 3: "Third"})
df.head()
```

## Video 2

Talking point: This is where the dataset gets compelling. .groupby("Sex")["Survived"].mean() reveals a striking real pattern: women survived at a dramatically higher rate than men — a direct reflection of the historical "women and children first" evacuation policy. Combining Pclass and Sex in one groupby shows an even sharper story: first-class women had the highest survival rate, third-class men the lowest. This single line of code encodes a piece of real history.



```python
df.describe()     # Show some overall statistics : mean, std, quartiles for numeric columns
```

#### How many survived per gender

```python
df.groupby("Sex")["Survived"].mean()
```

#### How many survived per Class

```python
df.groupby("Pclass_label")["Survived"].mean().sort_values(ascending=False)
```

#### Surevival rate per gender and Class

```python
df.groupby(["Pclass_label", "Sex"]).agg(
    survival_rate=("Survived", "mean"),
    avg_fare=("Fare", "mean"),
    n_passengers=("PassengerId", "count")
)
```

Talking point: Ask students to interpret the sign and strength of each correlation before revealing the answer:

Fare correlates positively with survival (wealthier passengers had better cabin access to lifeboats).

Pclass correlates negatively (class 1 = 1, class 3 = 3, so a negative correlation with survival means higher class number → lower survival, i.e., third class fared worst).



```python
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]

df[numeric_cols].corr()["Survived"].sort_values(ascending=False)
```
