#### How fast in Numpy?

```python
import numpy as np
import time

N = 10_000_000
data_list = list(range(N))
data_np = np.arange(N)
```

#### Run native python

```python
def python_sum(lst):
    total = 0
    for x in lst:
        total += x
    return total
```

```python
start = time.perf_counter()
py_result = python_sum(data_list)
py_time = time.perf_counter() - start
```

#### use Numpy

```python
start = time.perf_counter()
np_result = np.sum(data_np)
np_time = time.perf_counter() - start
```

```python
print(f"Pure Python sum: {py_time:.4f} s")
print(f"NumPy sum:       {np_time:.6f} s")
print(f"NumPy speedup:   {py_time/np_time:.1f}x faster")
```

#### An image is an array

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

# Load the image into a NumPy array
photo = mpimg.imread("horses.jpeg")   

# Inspect its structure
print("Shape:", photo.shape)   # (height, width, channels)
print("Dtype:", photo.dtype)   # usually uint8, values 0-255
print("Type:", type(photo))    # numpy.ndarray

# Display it
plt.imshow(photo)
plt.axis("off")
```

```python
SEPIA = np.array([[0.393, 0.769, 0.189],
                   [0.349, 0.686, 0.168],
                   [0.272, 0.534, 0.131]], dtype=np.float32)
```

```python
SEPIA.shape
```

```python
photo[:,:,:,np.newaxis].shape
```

```python
new_image = SEPIA.reshape((1,1,3,3)) @ photo[:,:,:,np.newaxis] /255
plt.imshow(new_image[:,:,:,0].clip(0,1))
```

## solutions..

1. Define new matrix to multiply by the image to generate the origical image

```python
no_color_change = np.eye(3)
no_color_change
```

```python
new=no_color_change.reshape((1,1,3,3))@ photo[:,:,:,np.newaxis]/255
plt.imshow(new[:,:,:,0].clip(0,1))
```

2. Define new matrix to multiply by the image to generate a gray-scale image

```python
Average = np.array([[0.333, 0.333, 0.333],
                   [0.333, 0.333, 0.333],
                   [0.333, 0.333, 0.333]], dtype=np.float32)
```

```python
new=Average.reshape((1,1,3,3))@ photo[:,:,:,np.newaxis]/255
plt.imshow(new[:,:,:,0].clip(0,1))
```

```python
plt.imshow(photo[:,:,:].mean(axis=2), cmap="gray")
```

3. Flip the image sideways

```python
plt.imshow(photo[:,::-1,:])
```

```python
plt.imshow(photo[::-1,:,:])   ## And upside downs
```

4. Change the order of the horses

```python
left_horse=photo[:,0:910,:] 
right_horse=photo[:,1600:2500,:]
mid_horse=photo[:,911:1599,:]

plt.imshow(np.hstack([mid_horse, left_horse, right_horse]))
```

5. transpose the image

```python
plt.imshow(photo.transpose((1,0,2)))
```
