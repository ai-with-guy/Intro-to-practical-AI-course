## Step 4: training data versus unseen data
A held-out test set distinguishes memorization from generalization.

```python
def split_data(data, test_size=0.3):
    return train_test_split(
        data, test_size=test_size, stratify=data["class"], random_state=42
    )
```

```python
from sklearn.model_selection import train_test_split
```

```python
spiral_train, spiral_test = split_data(pd.read_csv("data3_spiral.csv"))
utils.display_points(spiral_train, "Dataset 3: noisy intertwined spirals", spiral_test)
```

```python
spiral_tree = DecisionTreeClassifier(random_state=42)
spiral_tree.fit(spiral_train[FEATURES_2D], spiral_train["class"])
```

```python
tree_train_predictions = spiral_tree.predict(spiral_train[FEATURES_2D])
tree_test_predictions = spiral_tree.predict(spiral_test[FEATURES_2D])
print(f"training accuracy: {utils.accuracy_score(spiral_train['class'], tree_train_predictions):.3f}")
print(f"accuracy on unseen test data: {utils.accuracy_score(spiral_test['class'], tree_test_predictions):.3f}")
```

```python
utils.display_classifications(spiral_tree, spiral_train, "One tree: perfect memory, brittle boundary", test=spiral_test)
```

```python
utils.display_classifications(spiral_forest, spiral_train, "Random forest: averaging smooths unstable cuts", test=spiral_test)
```
