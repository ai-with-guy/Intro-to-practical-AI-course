## Reading the result

Each patch is a token. The heatmap shows one row of the selected layer's attention matrix: the selected patch is the **query**, and every colored patch is a possible **key**. Attention is directional, so clicking patch A and looking at B is not necessarily the same as clicking B and looking at A.

Early layers often focus on nearby texture and edges. Deeper layers can connect distant parts of an object. Individual heads may specialize, while **Mean** gives a simpler overview. These are raw attention weights, not a complete explanation of the model's prediction.
